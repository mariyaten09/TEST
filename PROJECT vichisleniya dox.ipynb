```python
users_1=df[df["date"]=="2023-11-01"]["user_id"].unique()
users_2=df[df["date"]=="2023-11-02"]["user_id"].unique()
returned=len(set(users_1)&set(users_2))
retention=returned / len(users_1) * 100
print("Пользователей 1 ноября:", len(users_1))
print("Вернулись 2 ноября:", returned)
print("Retention:", retention, "%")
```

```python
all_users=df["user_id"].nunique()
view_users=df[df["view_adverts"] > 0]["user_id"].nunique()
view_users=df[df["view_adverts"] > 0]["user_id"].nunique()
conversion=view_users / all_users * 100
print("Всего пользователей:", all_users)
print("Посмотрели объявления:", view_users)
print("Конверсия:", conversion, "%")
```
